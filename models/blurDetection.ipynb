```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch.nn as nn
from torch import DataLoader, sampler, random_split
from torchvision import datasets, models, transforms
```

```python
img_transforms = {
    "train":
    transforms.Compose([
        transforms.Grayscale(),
        transforms.RandomResizedCrop(size = 256),
        transforms.RandomRotation(degrees = 15),
        transforms.ColorJitter(),
        transforms.RandomHorizontalFlip(),
        transforms.CenterCrop(size = 224),
        transforms.ToTensor(),
        transforms.Normalize([0.5,],
                             [0.5,])
    ]),
    "test":
    transforms.Compose([
        transforms.Grayscale(),
        transforms.Resize(size = 256),
        transforms.CenterCrop(size = 224),
        transforms.ToTensor(),
        transforms.Normalize([0.5,],
                             [0.5,])
    ])
}
```

```python
# Loading the dataset

data = datasets.ImageFolder(r"C:\YaadChitra\models\blurDataset", transform=img_transforms["train"])
```

```python
# Splitting the dataset into training and validation sets (train-70%, validation-15%, test-15%)
train_size = int(0.7 * len(data))
val_size = int(0.15 * len(data))
test_size = len(data) - train_size - val_size

print("Train Data Length :", train_size)
print("Validation Data Length :", val_size)
print("Test Data Length :", test_size)
```

```python
# Splitting the dataset into training, validation, and test sets
train_ds, val_ds, test_ds = random_split(data, [train_size, val_size, test_size])
```

```python
# Applying the specified transformations to the training, validation, and test datasets.
train_ds.dataset.transform = img_transforms['train']
val_ds.dataset.transform = img_transforms['test']
test_ds.dataset.transform = img_transforms['test']

print("Length of train data: ", len(train_ds), "\n",
      "Length of val data:",  len(val_ds), "\n",
      "Length of test data:", len(test_ds))
```

```python
batch=32
train_Loader = DataLoader(train_ds, batch_size=batch, shuffle=True)
val_Loader = DataLoader(val_ds, batch_size=batch, shuffle=True)
test_Loader = DataLoader(test_ds, batch_size=batch, shuffle=True)
```
